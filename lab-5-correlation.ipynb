```python
#imports and packages
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
```

```python
# load dataset
df = pd.read_csv(r'dataset\current-power.csv', delimiter=",")
df
```

```python
df.describe()
```

```python
df.cov()
```

```python
x = df['Current']
y = df['Power']

plt.figure()
plt.scatter(x,y)
plt.title('Device Current vs Power Response')
plt.xlabel('Current (A)')
plt.ylabel('Power (W)')
plt.show()
```

```python
df.corr()
```

```python
slope, intercept, r_value, p_value, std_err = stats.linregress(x,y)
line = slope * x + intercept

plt.figure(figsize=(7,4))
plt.scatter(x, y, color = "#000000", alpha=0.6 ,label='Data Points')

plt.plot(x, line, color = "#653F02", label=f'Regression line (r = {r_value:.3f})')

plt.title('Device Current vs Power Response')
plt.xlabel('Current (A)')
plt.ylabel('Power (W)')
plt.legend()
plt.show()
```

# Air Quality

```python
air = pd.read_csv(r"dataset\air-quality.csv", delimiter=",")
air
```

# NOx-Humidity

```python
x = air['NOx']
y = air['Relative_Humidity']
slope, intercept, r_value, p_value, std_err = stats.linregress(x,y)
line = slope * x + intercept

plt.figure(figsize=(7,4))
plt.scatter(x, y, color = "#000000", alpha=0.6 ,label='Data Points')

plt.plot(x, line, color = "#653F02", label=f'Regression line (r = {r_value:.3f})')

plt.title('NOx vs Relative Humidity')
plt.xlabel('NOx (ug/cu.m)')
plt.ylabel('Relative Humidity (%)')
plt.legend()
plt.show()
```

# NO2-Humidity

```python
x = air['NO2']
y = air['Relative_Humidity']

slope, intercept, r_value, p_value, std_err = stats.linregress(x,y)
line = slope * x + intercept

plt.figure(figsize=(7,4))
plt.scatter(x, y, color = "#000000", alpha=0.6 ,label='Data Points')

plt.plot(x, line, color = "#653F02", label=f'Regression line (r = {r_value:.3f})')

plt.title('NO2 vs Relative Humidity Correlation')
plt.xlabel('NO2 (ug/cu.m)')
plt.ylabel('Relative Humidity (%)')
plt.legend()
plt.show()
```

# Temperature-Humidity

```python
x = air['Temperature']
y = air['Relative_Humidity']

slope, intercept, r_value, p_value, std_err = stats.linregress(x,y)
line = slope * x + intercept

plt.figure(figsize=(7,4))
plt.scatter(x, y, color = "#000000", alpha=0.6 ,label='Data Points')

plt.plot(x, line, color = "#653F02", label=f'Regression line (r = {r_value:.3f})')

plt.title('Temperature vs Relative Humidity Correlation')
plt.xlabel('Temp (C)')
plt.ylabel('Relative Humidity (%)')
plt.legend()
plt.show()
```
